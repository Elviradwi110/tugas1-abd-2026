```python
import polars as pl
import duckdb
import plotly
import os

print("Polars :", pl.__version__)
print("DuckDB :", duckdb.__version__)
print("Plotly :", plotly.__version__)
```

```python
DATA_PATH = "../data/raw/indonesia_palm_oil_v1_2_5.csv"

print("Dataset:", DATA_PATH)
print("File ditemukan:", os.path.exists(DATA_PATH))
```

```python
df = pl.scan_csv(DATA_PATH)

print("Dataset berhasil dibaca menggunakan Polars Lazy API")
```

```python
schema = df.collect_schema()

print("Jumlah kolom:", len(schema))
print("\nSchema:")
print(schema)
```

```python
row_count = df.select(
    pl.len().alias("jumlah_baris")
).collect()

print(row_count)
```

```python
from pathlib import Path

file_path = Path(DATA_PATH)

size_bytes = file_path.stat().st_size
size_mb = size_bytes / (1024 ** 2)
size_gb = size_bytes / (1024 ** 3)

print(f"Ukuran file : {size_mb:.2f} MB")
print(f"Ukuran file : {size_gb:.2f} GB")
```

```python
missing = df.select([
    pl.col(col).null_count().alias(col)
    for col in df.collect_schema().names()
]).collect()

missing
```

```python
total_rows = df.select(
    pl.len()
).collect().item()

unique_rows = df.unique().select(
    pl.len()
).collect().item()

duplicate_rows = total_rows - unique_rows

print("Total baris      :", total_rows)
print("Baris unik       :", unique_rows)
print("Duplikasi        :", duplicate_rows)
```

```python
missing_profile = (
    df
    .select([
        pl.col(col).null_count().alias(col)
        for col in df.collect_schema().names()
    ])
    .collect()
    .transpose(
        include_header=True,
        header_name="column",
        column_names=["missing_count"]
    )
    .with_columns(
        (
            pl.col("missing_count") / total_rows * 100
        ).round(2).alias("missing_percentage")
    )
    .sort("missing_count", descending=True)
)

missing_profile
```

```python
unique_rows = (
    df
    .unique()
    .select(pl.len().alias("unique_rows"))
    .collect()
    .item()
)

duplicate_rows = total_rows - unique_rows

print("Total baris   :", total_rows)
print("Baris unik    :", unique_rows)
print("Baris duplikat:", duplicate_rows)
```

```python
numeric_columns = [
    name
    for name, dtype in df.collect_schema().items()
    if dtype.is_numeric()
]

print("Kolom numerik:")
for col in numeric_columns:
    print("-", col)
```

```python
numeric_summary = df.select(numeric_columns).describe()

numeric_summary
```
